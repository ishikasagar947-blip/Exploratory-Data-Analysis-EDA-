##Exploratory Data Analysis on Cleaned Dataset

##Load Cleaned Dataset

```python
import pandas as pd
from google.colab import files

# Upload cleaned file
uploaded = files.upload()

# Load dataset
df = pd.read_csv("Cola_Cleaned.csv")
df.head()
```

##Descriptive Statistics

```python
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

##Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram
sns.histplot(df['value'], bins=30, kde=True)
plt.title("Value Distribution")
plt.show()
```

##### Observation 1
- Value ka distribution right‑skewed hai.
- Zyada entries lower range me clustered hain.


```python


# Line Plot
sns.lineplot(x='year', y='value', data=df)
plt.title("Trend Over Years")
plt.show()


```

##### Observation 2
- Year ke saath value me steady growth dikh rahi hai.
- 2015 ke baad decline noticeable hai.


```python
# Boxplot
sns.boxplot(x=df['value'])
plt.title("Outlier Detection")
plt.show()
```

##### Observation 3
- Outliers exist in value column.
- Kuch extreme high values dataset me present hain.


```python
# Bar Chart
sns.barplot(x='year', y='value', data=df)
plt.title("Yearly Comparison")
plt.show()

```

##### Observation 4
- Certain categories/years consistently higher values show karte hain.
- Category A sabse zyada contribute karti hai.


```python
# Heatmap
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

##### Observation 5
- Correlation matrix me strong relation between variables.
- Sales aur Profit positively correlated hain.


## Conclusion
- This Exploratory Data Analysis helped us understand the important patterns in the dataset.  
- We identified the distribution and detected outliers.  
- Trends over time and category comparisons became clear.  
- The correlation matrix revealed relationships between variables.  
- These insights will be useful for future analysis and decision-making.
