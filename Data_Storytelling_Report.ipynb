## Exploratory Data Analysis on Cleaned Cola Dataset


#### Methodology
- Dataset: Cleaned Cola dataset (Week 1 output)
- Environment: Google Colab
- Tools: Python, pandas, NumPy, Matplotlib, Seaborn
- Steps: Data cleaning → Exploratory Data Analysis → Storytelling Report


```python
import pandas as pd
import io

df = pd.read_csv(io.BytesIO(uploaded['Cola_Cleaned.csv']))

# Convert relevant columns to numeric types
df['value'] = pd.to_numeric(df['value'], errors='coerce')
df['year'] = pd.to_numeric(df['year'], errors='coerce').astype('Int64') # Using Int64 to allow for NaN if coercion fails
df['in_million_usd'] = pd.to_numeric(df['in_million_usd'], errors='coerce')

# Shape of dataset
df.shape

# Info about columns
df.info()

# Missing values
df.isnull().sum()

# Duplicate rows
df.duplicated().sum()
```

#### Data Quality Notes
- Dataset has X rows and Y columns
- No missing values / handled missing values
- Duplicates removed: Z
- Outliers identified in column 'Value'

##Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram
df['value'].hist()
plt.title("Histogram of Value")
plt.show()

# Boxplot
sns.boxplot(x=df['value'])
plt.title("Boxplot of Value")
plt.show()

# Line plot (example with Year column)
df.groupby('year')['value'].sum().plot(kind='line')
plt.title("Trend over Years")
plt.show()

# Bar chart (assuming a 'Category' column, which might not exist in the current df)
# If 'Category' doesn't exist, this will cause a KeyError.
# For now, let's comment it out or adapt it if a suitable categorical column is identified.
# df['Category'].value_counts().plot(kind='bar')
# plt.title("Category Comparison")
# plt.show()

# Heatmap (requires numeric data for correlation, and current df is empty/object dtypes)
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
